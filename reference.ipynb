```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy import text
import pandas as pd;
```

```python
load_dotenv()

DB_USER = os.getenv("DB_USER");
DB_PASSWORD = os.getenv("DB_PASSWORD");
DB_HOST = os.getenv("DB_HOST");
DB_PORT = os.getenv("DB_PORT", 3306);
DB_SIAKAD = os.getenv("DB_SIAKAD");

siakadConnection = f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_SIAKAD}"
siakadEngine = create_engine(siakadConnection)
```

```python
semesterQuery = text("""
SELECT
    id,
    name,
    code,
    semester,
    year_of_study
FROM semesters
WHERE
    year_of_study >= 2018 AND
    year_of_study <= 2026 AND
    NOT semester = 3
""")

semesterDf = pd.read_sql(semesterQuery, siakadEngine)

semesterDf
```

```python
semesterDf.to_csv('data/referensi/semester.csv', index=False)
```
