### Connection Mysql

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy import text
```

```python
load_dotenv()

DB_USER = os.getenv("DB_USER");
DB_PASSWORD = os.getenv("DB_PASSWORD");
DB_HOST = os.getenv("DB_HOST");
DB_PORT = os.getenv("DB_PORT", 3306);
DB_SIAKAD = os.getenv("DB_SIAKAD");

siakadConnection = f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_SIAKAD}"
siakadEngine = create_engine(siakadConnection)
```

### Library

```python
import pandas as pd
```

### Jabfung

```python
query = text("""
SELECT
	d.NAME AS dosen,
	jabfung.NAME AS jabfung,
	jabfung.start_date,
	s.CODE AS semester_code,
	s.year_of_study 
FROM
	dosen_functional_position_histories AS jabfung
	JOIN dosen_list AS d ON jabfung.dosen_uuid = d.uuid
	LEFT JOIN semesters AS s ON jabfung.start_date >= s.start_date 
	AND jabfung.start_date <= s.end_date 
	AND s.semester IN ( 1, 2 ) 
	AND s.year_of_study
""")
```

```python
df = pd.read_sql(query, siakadEngine)
```

```python
df['year_of_study'] = df['year_of_study'].astype('Int64').astype(str)
```

```python
df.to_csv('data/dosen/jabatan-fungsional.csv', index=False)
```
