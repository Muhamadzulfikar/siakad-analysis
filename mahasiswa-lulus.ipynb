### Connection MYSQL

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy import text
```

```python
load_dotenv()

DB_USER = os.getenv("DB_USER");
DB_PASSWORD = os.getenv("DB_PASSWORD");
DB_HOST = os.getenv("DB_HOST");
DB_PORT = os.getenv("DB_PORT", 3306);
DB_SIAKAD = os.getenv("DB_SIAKAD");

connection = f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_SIAKAD}"
engine = create_engine(connection)
```

### Library

```python
import pandas as pd;
```

### Mahasiswa Lulus

```python
query = text("""
SELECT
	registers.semester_code AS register_semester_code,
	keluar.semester_code AS keluar_semester_code,
	registers.nim,
	semesters.year_of_study,
	prodi.name AS program_study

FROM
	mahasiswa_keluar AS keluar
	
    JOIN mahasiswa_keluar_details AS keluar_details ON keluar_details.keluar_id = keluar.id
	
    JOIN mahasiswa_registers AS registers ON keluar_details.mahasiswa_register_uuid = registers.registration_uuid
	
    JOIN prodi ON registers.prodi_uuid = prodi.uuid
    
    JOIN semesters ON keluar.semester_code = semesters.code

WHERE
    keluar.jeniskeluar_code = 1
""")
```

```python
df = pd.read_sql(query, engine)
```

```python
df.to_csv('data/mahasiswa/mahasiswa-lulus.csv', index=False)
```
